```python
import pandas as pd
import matplotlib.pyplot as plt

df = pd.read_csv("NetFlix.csv")

# Clean comma-separated genres
genres = df['genres'].dropna().str.split(', ').explode()
genre_counts = genres.value_counts()

top_10_genres = genre_counts.head(10)
bottom_10_genres = genre_counts.tail(10)
most_popular = genre_counts.idxmax()
least_popular = genre_counts.idxmin()

print("--- Top 10 Genres ---")
print(top_10_genres)

print("\n--- Bottom 10 Genres ---")
print(bottom_10_genres)

print(f"\nMost Popular Genre: {most_popular} ({genre_counts.max()} titles)")
print(f"Least Popular Genre: {least_popular} ({genre_counts.min()} titles)")

# Visualization
plt.figure(figsize=(10, 5))
top_10_genres.plot(kind='bar', color='skyblue', edgecolor='black')
plt.title('Top 10 Netflix Genres')
plt.xlabel('Genre')
plt.ylabel('Number of Titles')
plt.xticks(rotation=45, ha='right')
plt.tight_layout()
plt.show()
```

```python
import pandas as pd

df = pd.read_csv("NetFlix.csv")

# Filter by type
movies = df[df['type'].str.strip() == 'Movie']
tv_shows = df[df['type'].str.strip() == 'TV Show']

oldest_movie = movies[movies['release_year'] == movies['release_year'].min()]
newest_movie = movies[movies['release_year'] == movies['release_year'].max()]

oldest_tv = tv_shows[tv_shows['release_year'] == tv_shows['release_year'].min()]
newest_tv = tv_shows[tv_shows['release_year'] == tv_shows['release_year'].max()]

print("=== OLDEST MOVIE ===")
print(oldest_movie[['title', 'type', 'release_year', 'genres']].to_string(index=False))

print("\n=== NEWEST MOVIE ===")
print(newest_movie[['title', 'type', 'release_year', 'genres']].head(1).to_string(index=False))

print("\n=== OLDEST TV SHOW ===")
print(oldest_tv[['title', 'type', 'release_year', 'genres']].to_string(index=False))

print("\n=== NEWEST TV SHOW ===")
print(newest_tv[['title', 'type', 'release_year', 'genres']].head(1).to_string(index=False))
```

```python
import pandas as pd
import matplotlib.pyplot as plt

df = pd.read_csv("NetFlix.csv")

countries = df['country'].dropna().str.split(', ').explode()
country_counts = countries.value_counts()

top_country = country_counts.idxmax()
top_10_countries = country_counts.head(10)
bottom_10_countries = country_counts.tail(10)

print(f"Country with Highest Content: {top_country} ({country_counts.max()} titles)\n")

print("--- Top 10 Countries ---")
print(top_10_countries)

print("\n--- Bottom 10 Countries ---")
print(bottom_10_countries)

# Horizontal Bar Chart
plt.figure(figsize=(10, 6))
top_10_countries.sort_values(ascending=True).plot(kind='barh', color='crimson')
plt.title('Top 10 Countries by Netflix Content')
plt.xlabel('Number of Titles')
plt.ylabel('Country')
plt.tight_layout()
plt.show()
```

```python
import pandas as pd
import matplotlib.pyplot as plt

df = pd.read_csv("NetFlix.csv")

# Directors by Movies
movies_df = df[(df['type'].str.strip() == 'Movie') & (df['director'].notna())]
movie_directors = movies_df['director'].str.split(', ').explode().value_counts()
top_movie_dir = movie_directors.idxmax()

# Directors by TV Shows
tv_df = df[(df['type'].str.strip() == 'TV Show') & (df['director'].notna())]
tv_directors = tv_df['director'].str.split(', ').explode().value_counts()
top_tv_dir = tv_directors.idxmax()

# Top 10 Directors Overall
all_directors = df['director'].dropna().str.split(', ').explode().value_counts()
top_10_directors = all_directors.head(10)

print(f"Top Movie Director: {top_movie_dir} ({movie_directors.max()} movies)")
print(f"Top TV Show Director: {top_tv_dir} ({tv_directors.max()} TV shows)\n")

print("--- Top 10 Directors Overall ---")
print(top_10_directors)

# Bar Chart
plt.figure(figsize=(10, 5))
top_10_directors.plot(kind='bar', color='teal', edgecolor='black')
plt.title('Top 10 Directors on Netflix')
plt.xlabel('Director')
plt.ylabel('Number of Titles')
plt.xticks(rotation=45, ha='right')
plt.tight_layout()
plt.show()
```

```python
import pandas as pd
import matplotlib.pyplot as plt

df = pd.read_csv("NetFlix.csv")

before_2020 = (df['release_year'] < 2020).sum()
during_2020 = (df['release_year'] == 2020).sum()
after_2020 = (df['release_year'] > 2020).sum()

comparison = pd.Series({
    'Before 2020': before_2020,
    'During 2020': during_2020,
    'After 2020': after_2020
})

print("--- Netflix Releases Comparison ---")
print(comparison)

# Bar Chart
plt.figure(figsize=(8, 5))
bars = plt.bar(comparison.index, comparison.values, color=['#4C72B0', '#DD8452', '#55A868'], edgecolor='black')

for bar in bars:
    yval = bar.get_height()
    plt.text(bar.get_x() + bar.get_width()/2, yval + 20, f'{yval:,}', ha='center', va='bottom', fontweight='bold')

plt.title('Netflix Releases: Before, During, and After 2020')
plt.xlabel('Time Period')
plt.ylabel('Number of Releases')
plt.ylim(0, max(comparison.values) * 1.15)
plt.tight_layout()
plt.show()
```
