```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
from sklearn.model_selection import train_test_split
from sklearn.linear_model import LinearRegression
from sklearn.metrics import mean_squared_error,r2_score
```

```python
df =pd.read_csv("housing.csv")
```

```python
x =df[["median_income"]]
y=df["median_house_value"]
```

```python
X_train,X_test, y_train, y_test= train_test_split(
    x,
    y,
    test_size=0.2,
    random_state=42
)
```

```python
model = LinearRegression()
model.fit(X_train,y_train)
```

```python
print("Coefficient:",model.coef_[0])
print("Intercept:",model.intercept_)
```

```python
y_pred =model.predict(X_test)
```

```python

print("Predicted Values:")

print(y_pred[:10])
```

```python
mse = mean_squared_error(y_test, y_pred)

print("Mean Squared Error:", mse)
```

```python
r2 = r2_score(y_test, y_pred)

print("R2 Score:", r2)
```

```python
import pandas as pd

# Load dataset
df = pd.read_csv("housing.csv")

# 1. Complete Correlation Matrix
correlation = df.corr(numeric_only=True)
print("=== Complete Correlation Matrix ===")
print(correlation)

# 2. Identify Strongest & Weakest Correlations (excluding self-correlation = 1.0)
corr_pairs = correlation.unstack()
corr_pairs = corr_pairs[corr_pairs < 1.0]

# Strongest Positive
strongest_pos = corr_pairs.idxmax()
pos_val = corr_pairs.max()

# Strongest Negative
strongest_neg = corr_pairs.idxmin()
neg_val = corr_pairs.min()

# Weakest Correlation (closest to 0)
weakest = corr_pairs.abs().idxmin()
weak_val = corr_pairs[weakest]

print("\n=== Key Correlation Insights ===")
print(f"Strongest Positive Correlation: {strongest_pos} = {pos_val:.4f}")
print(f"Strongest Negative Correlation: {strongest_neg} = {neg_val:.4f}")
print(f"Weakest Correlation: {weakest} = {weak_val:.4f}")
```

```python
import pandas as pd
from sklearn.linear_model import LinearRegression
from sklearn.metrics import mean_squared_error, r2_score
from sklearn.model_selection import train_test_split

# Load dataset
df = pd.read_csv("housing.csv")

# Target Variable
y = df["median_house_value"]

# --- 1. Multiple Linear Regression Model ---
features = ["median_income", "housing_median_age", "total_rooms", "population", "households"]
X_mult = df[features]

X_train_m, X_test_m, y_train_m, y_test_m = train_test_split(
    X_mult, y, test_size=0.2, random_state=42
)

model_mult = LinearRegression()
model_mult.fit(X_train_m, y_train_m)
y_pred_m = model_mult.predict(X_test_m)

r2_mult = r2_score(y_test_m, y_pred_m)
mse_mult = mean_squared_error(y_test_m, y_pred_m)

# --- 2. Simple Linear Regression Model (median_income only) ---
X_simple = df[["median_income"]]

X_train_s, X_test_s, y_train_s, y_test_s = train_test_split(
    X_simple, y, test_size=0.2, random_state=42
)

model_simple = LinearRegression()
model_simple.fit(X_train_s, y_train_s)
y_pred_s = model_simple.predict(X_test_s)

r2_simple = r2_score(y_test_s, y_pred_s)
mse_simple = mean_squared_error(y_test_s, y_pred_s)

# Display & Compare Results
print("=== Multiple Linear Regression Model ===")
print("R2 Score:", r2_mult)
print("Mean Squared Error (MSE):", mse_mult)

print("\n=== Simple Linear Regression Model ===")
print("R2 Score:", r2_simple)
print("Mean Squared Error (MSE):", mse_simple)
```
