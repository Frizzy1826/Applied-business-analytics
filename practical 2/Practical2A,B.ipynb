```python
import pandas as pd

# Load the dataset containing the specified columns
df = pd.read_csv("ecomer.csv")
print("First 5 rows of the dataset:")
print(df.head())

# Extracting the 'revenue' column for statistical analysis
sales = df["revenue"]
```

```python
print("--- Individual Statistics ---")
print("Mean :", sales.mean())
print("Median :", sales.median())
print("Minimum :", sales.min())
print("Maximum :", sales.max())
print("Range :", sales.max() - sales.min())
print("Variance :", sales.var())
print("Standard Deviation :", sales.std())
print("Count :", sales.count())
print("Sum :", sales.sum())
```

```python
print("\n--- Descriptive Statistics Summary ---")
print(sales.describe())
```

```python
# B] Visualizations
import matplotlib.pyplot as plt
import seaborn as sns

# 1. Histogram
plt.figure(figsize=(8,5))
plt.hist(sales, bins=10, color="skyblue", edgecolor="black")
plt.title("Histogram of Revenue Bhavesh T039")
plt.xlabel("Revenue")
plt.ylabel("Frequency")
plt.show()

# 2. Box Plot
plt.figure(figsize=(8,3))
sns.boxplot(x=sales, color="orange")
plt.title("Box Plot of Revenue Bhavesh T039")
plt.xlabel("Revenue")
plt.show()
```
